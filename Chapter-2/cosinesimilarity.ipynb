```python
import numpy as np
import matplotlib.pyplot as plt

# Define vectors
u = np.array([1, 2, 3])
v = np.array([2, 4, 6])  # v = 2u

# Compute dot product
dot_product = np.dot(u, v)

# Compute magnitudes
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)

# Compute cosine similarity and distance
cosine_similarity = dot_product / (norm_u * norm_v)
cosine_distance = 1 - cosine_similarity

# Print values
print(f"u · v = {dot_product}")
print(f"||u|| = {norm_u:.2f}")
print(f"||v|| = {norm_v:.2f}")
print(f"Cosine similarity = {cosine_similarity:.2f}")
print(f"Cosine distance = {cosine_distance:.2f}")

# 3D Visualization
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(8, 7))
ax = fig.add_subplot(111, projection='3d')

# Plot vectors
ax.quiver(0, 0, 0, u[0], u[1], u[2], color='blue', linewidth=3, label='u')
ax.quiver(0, 0, 0, v[0], v[1], v[2], color='red', linewidth=3, label='v')

# Annotate vectors
ax.text(u[0], u[1], u[2], 'u', color='blue', fontsize=13, weight='bold')
ax.text(v[0], v[1], v[2], 'v', color='red', fontsize=13, weight='bold')

# Annotate similarity and distance
ax.text2D(0.05, 0.92, f"Cosine similarity = {cosine_similarity:.2f}\nCosine distance = {cosine_distance:.2f}", 
          transform=ax.transAxes, fontsize=13, color='black', bbox=dict(facecolor='white', alpha=0.7))

# Set axes
ax.set_xlim([0, 7])
ax.set_ylim([0, 7])
ax.set_zlim([0, 7])
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Cosine Similarity and Distance\nu and v point in the same direction')
```
