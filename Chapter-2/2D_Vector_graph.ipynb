```python
import matplotlib.pyplot as plt
import numpy as np

# Define the vector
v = np.array([3, 4])

# Calculate magnitude
magnitude = np.sqrt(v[0]**2 + v[1]**2)

# Create the plot
fig, ax = plt.subplots(1, 1, figsize=(8, 6))

# Plot the vector as an arrow from origin
ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, 
          color='blue', width=0.005, label=f'Vector v = ({v[0]}, {v[1]})')

# Plot the magnitude as a line from origin to vector tip
ax.plot([0, v[0]], [0, v[1]], 'r--', linewidth=2, 
        label=f'Magnitude ||v|| = {magnitude}')

# Mark the vector endpoint
ax.plot(v[0], v[1], 'ro', markersize=8, label=f'Point ({v[0]}, {v[1]})')

# Mark the origin
ax.plot(0, 0, 'ko', markersize=8, label='Origin (0, 0)')

# Set up the grid and axes
ax.grid(True, alpha=0.3)
ax.set_xlim(-1, 5)
ax.set_ylim(-1, 5)
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_title(f'Vector v = (3, 4) with Magnitude ||v|| = {magnitude}')

# Add annotations
ax.annotate(f'v = ({v[0]}, {v[1]})', xy=(v[0], v[1]), xytext=(v[0]+0.2, v[1]+0.2),
            fontsize=12, color='blue')
ax.annotate(f'||v|| = {magnitude}', xy=(v[0]/2, v[1]/2), xytext=(v[0]/2-0.5, v[1]/2+0.3),
            fontsize=12, color='red')

# Add legend
ax.legend()

# Make axes equal to preserve vector proportions
ax.set_aspect('equal')

# Show the plot
plt.tight_layout()
plt.show()

# Print calculation details
print(f"Vector v = ({v[0]}, {v[1]})")
print(f"Magnitude calculation:")
print(f"||v|| = √({v[0]}² + {v[1]}²)")
print(f"||v|| = √({v[0]**2} + {v[1]**2})")
print(f"||v|| = √{v[0]**2 + v[1]**2}")
print(f"||v|| = {magnitude}")
```
