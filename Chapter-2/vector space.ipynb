```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define vectors and their names
vectors = [
    np.array([1, 3, 0.5]),   # apple (fruit)
    np.array([1, 3, 1]),   # banana
    np.array([1, 3, 1.5]),   # orange
    np.array([4, 3, 1.5]),   # apple (company)
    np.array([4, 3, 1]),   # google
    np.array([4, 3, 0.5])    # amazon
]
names = [
    "apple", "banana", "orange",
    "apple", "google", "amazon"
]
groups = [
    "fruit", "fruit", "fruit",
    "company", "company", "company"
]
colors = ['red', 'orange', 'gold', 'blue', 'green', 'purple']

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot each vector with its group color and name
for v, name, group, color in zip(vectors, names, groups, colors):
    ax.quiver(0, 0, 0, v[0], v[1], v[2], color=color, arrow_length_ratio=0.1, linewidth=3)
    ax.text(v[0], v[1], v[2], name, fontsize=11, color=color, weight='bold')

# Custom legends for groups
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], color='red', lw=3, label='Fruit'),
    Line2D([0], [0], color='orange', lw=3, label='Fruit'),
    Line2D([0], [0], color='gold', lw=3, label='Fruit'),
    Line2D([0], [0], color='blue', lw=3, label='Company'),
    Line2D([0], [0], color='green', lw=3, label='Company'),
    Line2D([0], [0], color='purple', lw=3, label='Company'),
]
ax.legend(handles=legend_elements, labels=['Fruit', 'Company'], loc='upper left')

ax.set_xlim([0, 4])
ax.set_ylim([0, 4])
ax.set_zlim([0, 4])
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Vector Space: Fruits vs Companies')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
