```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

def create_3d_vector_graph():
    """
    Create a 3D vector visualization for vector w = [1, 2, 2] with magnitude 3
    """
    
    # Define the 3D vector
    w = np.array([1, 2, 2])
    
    # Calculate magnitude
    magnitude = np.sqrt(w[0]**2 + w[1]**2 + w[2]**2)
    
    print(f"Vector w = [{w[0]}, {w[1]}, {w[2]}]")
    print(f"Magnitude calculation:")
    print(f"||w|| = √({w[0]}² + {w[1]}² + {w[2]}²)")
    print(f"||w|| = √({w[0]**2} + {w[1]**2} + {w[2]**2})")
    print(f"||w|| = √{w[0]**2 + w[1]**2 + w[2]**2}")
    print(f"||w|| = {magnitude}")
    print("-" * 40)
    
    # Create 3D plot
    fig = plt.figure(figsize=(14, 10))
    ax = fig.add_subplot(111, projection='3d')
    
    # Plot the vector as an arrow from origin
    ax.quiver(0, 0, 0, w[0], w[1], w[2], 
              color='blue', arrow_length_ratio=0.15, linewidth=5,
              label=f'Vector w = [{w[0]}, {w[1]}, {w[2]}]')
    
    # Plot the magnitude as a line from origin to vector tip
    ax.plot([0, w[0]], [0, w[1]], [0, w[2]], 'r--', linewidth=3,
            label=f'Magnitude ||w|| = {magnitude}')
    
    # Mark the vector endpoint with a large marker
    ax.scatter(w[0], w[1], w[2], color='red', s=200, marker='o',
               label=f'Point ({w[0]}, {w[1]}, {w[2]})')
    
    # Add a LARGE marker specifically for z=2 visibility
    ax.scatter(w[0], w[1], w[2], color='yellow', s=300, marker='*',
               label=f'Z = {w[2]} Position', edgecolors='black', linewidth=2)
    
    # Mark the origin
    ax.scatter(0, 0, 0, color='black', s=150, label='Origin (0, 0, 0)')
    
    # Add coordinate projections with thicker lines
    # X-axis projection (red dotted)
    ax.plot([0, w[0]], [0, 0], [0, 0], 'r:', alpha=0.8, linewidth=3, label='X-component')
    # Y-axis projection (green dotted)
    ax.plot([w[0], w[0]], [0, w[1]], [0, 0], 'g:', alpha=0.8, linewidth=3, label='Y-component')
    # Z-axis projection (purple dotted) - THIS SHOWS Z=2!
    ax.plot([w[0], w[0]], [w[1], w[1]], [0, w[2]], 'm:', alpha=0.8, linewidth=4, label='Z-component = 2')
    
    # Add vertical line from XY-plane to show Z height
    ax.plot([w[0], w[0]], [w[1], w[1]], [0, w[2]], 'orange', alpha=0.9, linewidth=6)
    
    # Set labels and title
    ax.set_xlabel('X-axis', fontsize=14)
    ax.set_ylabel('Y-axis', fontsize=14)
    ax.set_zlabel('Z-axis', fontsize=14)
    ax.set_title(f'3D Vector w = [1, 2, 2] with Z-component = {w[2]}', 
                 fontsize=16, fontweight='bold')
    
    # Set axis limits with better spacing
    ax.set_xlim(0, 3.5)
    ax.set_ylim(0, 3.5)
    ax.set_zlim(0, 3.5)
    
    # Add grid
    ax.grid(True, alpha=0.3)
    
    # Add multiple annotations to highlight Z=2
    ax.text(w[0]+0.2, w[1]+0.2, w[2]+0.2, f'w = [{w[0]}, {w[1]}, {w[2]}]',
            fontsize=14, color='blue', fontweight='bold',
            bbox=dict(boxstyle="round,pad=0.3", facecolor="lightblue", alpha=0.7))
    
    # Add specific Z=2 annotation
    ax.text(w[0], w[1], w[2]+0.3, f'Z = {w[2]}',
            fontsize=16, color='red', fontweight='bold',
            bbox=dict(boxstyle="round,pad=0.3", facecolor="yellow", alpha=0.8))
    
    # Set a better viewing angle to see Z-axis clearly
    ax.view_init(elev=20, azim=45)  # Elevation=20°, Azimuth=45°
    
    # Add legend
    ax.legend(fontsize=11, loc='upper left')
    
    # Show the plot
    plt.tight_layout()
    plt.show()
    
    # Print confirmation
    print(f"\n✓ Arrow endpoint is at Z = {w[2]}")
    print(f"✓ Look for the YELLOW STAR and ORANGE line showing Z = {w[2]}")
```

```python
def main():
    """
    Main function to run the 3D vector visualization
    """
    print("3D Vector Visualization")
    print("=" * 50)
    print("Vector w = [1, 2, 2] with Magnitude ||w|| = 3")
    print("=" * 50)
    
    # Create 3D vector graph
    create_3d_vector_graph()
    
    print("\nCreating 2D projections...")
    # Create 2D projections
    create_2d_projections()
    
    print("\nVisualization complete!")

if __name__ == "__main__":
    main()
```
