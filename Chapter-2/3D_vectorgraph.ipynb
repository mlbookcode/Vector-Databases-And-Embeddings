```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

def create_3d_vector_graph():
    """
    Create a 3D vector visualization for vector w = [1, 2, 2] with magnitude 3
    """
    
    # Define the 3D vector
    w = np.array([1, 2, 2])
    
    # Calculate magnitude
    magnitude = np.sqrt(w[0]**2 + w[1]**2 + w[2]**2)
    
    print(f"Vector w = [{w[0]}, {w[1]}, {w[2]}]")
    print(f"Magnitude calculation:")
    print(f"||w|| = √({w[0]}² + {w[1]}² + {w[2]}²)")
    print(f"||w|| = √({w[0]**2} + {w[1]**2} + {w[2]**2})")
    print(f"||w|| = √{w[0]**2 + w[1]**2 + w[2]**2}")
    print(f"||w|| = {magnitude}")
    print("-" * 40)
    
    # Create 3D plot
    fig = plt.figure(figsize=(12, 9))
    ax = fig.add_subplot(111, projection='3d')
    
    # Plot the vector as an arrow from origin
    ax.quiver(0, 0, 0, w[0], w[1], w[2], 
              color='blue', arrow_length_ratio=0.1, linewidth=4,
              label=f'Vector w = [{w[0]}, {w[1]}, {w[2]}]')
    
    # Plot the magnitude as a line from origin to vector tip
    ax.plot([0, w[0]], [0, w[1]], [0, w[2]], 'r--', linewidth=3,
            label=f'Magnitude ||w|| = {magnitude}')
    
    # Mark the vector endpoint
    ax.scatter(w[0], w[1], w[2], color='red', s=150, 
               label=f'Point ({w[0]}, {w[1]}, {w[2]})')
    
    # Mark the origin
    ax.scatter(0, 0, 0, color='black', s=150, label='Origin (0, 0, 0)')
    
    # Add coordinate projections (dotted lines to show 3D structure)
    # X-axis projection
    ax.plot([0, w[0]], [0, 0], [0, 0], 'g:', alpha=0.7, linewidth=2)
    ax.plot([w[0], w[0]], [0, w[1]], [0, 0], 'g:', alpha=0.7, linewidth=2)
    ax.plot([w[0], w[0]], [w[1], w[1]], [0, w[2]], 'g:', alpha=0.7, linewidth=2)
    
    # Set labels and title
    ax.set_xlabel('X-axis', fontsize=12)
    ax.set_ylabel('Y-axis', fontsize=12)
    ax.set_zlabel('Z-axis', fontsize=12)
    ax.set_title(f'3D Vector w = [1, 2, 2] with Magnitude ||w|| = {magnitude}', 
                 fontsize=14, fontweight='bold')
    
    # Set axis limits
    ax.set_xlim(0, 3)
    ax.set_ylim(0, 3)
    ax.set_zlim(0, 3)
    
    # Add grid
    ax.grid(True, alpha=0.3)
    
    # Add annotations
    ax.text(w[0]+0.1, w[1]+0.1, w[2]+0.1, f'w = [{w[0]}, {w[1]}, {w[2]}]',
            fontsize=12, color='blue', fontweight='bold')
    
    # Add legend
    ax.legend(fontsize=10)
    
    # Show the plot
    plt.tight_layout()
    plt.show()

def create_2d_projections():
    """
    Create 2D projections of the 3D vector in X-Y, X-Z, and Y-Z planes
    """
    
    w = np.array([1, 2, 2])
    
    # Create subplots for projections
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    # X-Y projection
    axes[0].quiver(0, 0, w[0], w[1], angles='xy', scale_units='xy', scale=1, 
                   color='blue', width=0.008)
    axes[0].plot([0, w[0]], [0, w[1]], 'r--', alpha=0.7, linewidth=2)
    axes[0].scatter(w[0], w[1], color='red', s=80)
    axes[0].scatter(0, 0, color='black', s=80)
    axes[0].set_xlim(-0.5, 2.5)
    axes[0].set_ylim(-0.5, 2.5)
    axes[0].set_xlabel('X', fontsize=12)
    axes[0].set_ylabel('Y', fontsize=12)
    axes[0].set_title('X-Y Projection', fontsize=12, fontweight='bold')
    axes[0].grid(True, alpha=0.3)
    axes[0].set_aspect('equal')
    axes[0].annotate(f'({w[0]}, {w[1]})', xy=(w[0], w[1]), xytext=(w[0]+0.1, w[1]+0.1))
    
    # X-Z projection
    axes[1].quiver(0, 0, w[0], w[2], angles='xy', scale_units='xy', scale=1, 
                   color='blue', width=0.008)
    axes[1].plot([0, w[0]], [0, w[2]], 'r--', alpha=0.7, linewidth=2)
    axes[1].scatter(w[0], w[2], color='red', s=80)
    axes[1].scatter(0, 0, color='black', s=80)
    axes[1].set_xlim(-0.5, 2.5)
    axes[1].set_ylim(-0.5, 2.5)
    axes[1].set_xlabel('X', fontsize=12)
    axes[1].set_ylabel('Z', fontsize=12)
    axes[1].set_title('X-Z Projection', fontsize=12, fontweight='bold')
    axes[1].grid(True, alpha=0.3)
    axes[1].set_aspect('equal')
    axes[1].annotate(f'({w[0]}, {w[2]})', xy=(w[0], w[2]), xytext=(w[0]+0.1, w[2]+0.1))
    
    # Y-Z projection
    axes[2].quiver(0, 0, w[1], w[2], angles='xy', scale_units='xy', scale=1, 
                   color='blue', width=0.008)
    axes[2].plot([0, w[1]], [0, w[2]], 'r--', alpha=0.7, linewidth=2)
    axes[2].scatter(w[1], w[2], color='red', s=80)
    axes[2].scatter(0, 0, color='black', s=80)
    axes[2].set_xlim(-0.5, 2.5)
    axes[2].set_ylim(-0.5, 2.5)
    axes[2].set_xlabel('Y', fontsize=12)
    axes[2].set_ylabel('Z', fontsize=12)
    axes[2].set_title('Y-Z Projection', fontsize=12, fontweight='bold')
    axes[2].grid(True, alpha=0.3)
    axes[2].set_aspect('equal')
    axes[2].annotate(f'({w[1]}, {w[2]})', xy=(w[1], w[2]), xytext=(w[1]+0.1, w[2]+0.1))
    
    plt.tight_layout()
    plt.show()

def main():
    """
    Main function to run the 3D vector visualization
    """
    print("3D Vector Visualization")
    print("=" * 50)
    print("Vector w = [1, 2, 2] with Magnitude ||w|| = 3")
    print("=" * 50)
    
    # Create 3D vector graph
    create_3d_vector_graph()
    
    print("\nCreating 2D projections...")
    # Create 2D projections
    create_2d_projections()
    
    print("\nVisualization complete!")

if __name__ == "__main__":
    main()
```
