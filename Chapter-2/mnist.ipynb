```python
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml
import numpy as np

# Load the MNIST dataset
print("Loading MNIST dataset...")
mnist = fetch_openml('mnist_784', version=1, as_frame=False)
# 'mnist_784': 70,000 images of 28x28 pixels = 784 features per image

X = mnist.data                          # Feature matrix: shape (70000, 784)
y = mnist.target.astype(int)            # Labels: shape (70000,)

# Display information about the dataset
print(f"\nMNIST Dataset Information:")
print(f"  - Total samples: {X.shape[0]}")           # 70,000 images
print(f"  - Dimensions per image: {X.shape[1]}")    # 784 features (28x28 flattened)
print(f"  - Image shape: 28 x 28 pixels")
print(f"  - Pixel value range: {X.min():.0f} to {X.max():.0f}")
print(f"  - Number of classes: {len(np.unique(y))}")  # 10 digits (0-9)

# Select 16 random samples to display
n_samples = 16
random_indices = np.random.choice(X.shape[0], n_samples, replace=False)
# Choose 16 random indices from the dataset

# Create a 4x4 grid to display images
fig, axes = plt.subplots(4, 4, figsize=(10, 10))
# Create subplot grid: 4 rows x 4 columns

fig.suptitle(f'MNIST Dataset Visualization\nEach image: 28×28 pixels = 784 dimensions', 
             fontsize=16, fontweight='bold')
# Add main title showing dimensionality

# Plot each image
for idx, ax in enumerate(axes.flat):    # Loop through all 16 subplots
    # Get the image and label
    image_idx = random_indices[idx]     # Get random image index
    image = X[image_idx].reshape(28, 28)  # Reshape from (784,) to (28, 28)
    label = y[image_idx]                # Get the digit label
    
    # Display the image
    ax.imshow(image, cmap='gray')       # Show image in grayscale
    ax.set_title(f'Label: {label}', fontsize=12)  # Title shows the digit
    ax.axis('off')                      # Hide axis ticks and labels

plt.tight_layout()
plt.show()

# Display detailed information about one image
sample_idx = random_indices[0]
sample_image = X[sample_idx]
sample_label = y[sample_idx]

print(f"\n=== Detailed Image Information ===")
print(f"Sample index: {sample_idx}")
print(f"Label (digit): {sample_label}")
print(f"Original shape (flattened): {sample_image.shape}")  # (784,)
print(f"Reshaped to 2D: (28, 28)")
print(f"Total dimensions: 784 (28 × 28)")
print(f"Each dimension represents one pixel value")
print(f"Pixel values range: 0 (white) to 255 (black)")

# Visualize dimensionality concept
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Show one image in detail
sample_reshaped = sample_image.reshape(28, 28)
axes[0].imshow(sample_reshaped, cmap='gray')
axes[0].set_title(f'Original Image\nLabel: {sample_label}', fontsize=12, fontweight='bold')
axes[0].axis('off')

# Show flattened representation
axes[1].plot(sample_image, linewidth=0.5)
axes[1].set_title('Flattened to 784 Dimensions\n(Vector representation)', 
                  fontsize=12, fontweight='bold')
axes[1].set_xlabel('Dimension (pixel position)')
axes[1].set_ylabel('Pixel value')
axes[1].grid(True, alpha=0.3)
axes[1].set_xlim(0, 784)

# Show heatmap representation
im = axes[2].imshow(sample_reshaped, cmap='viridis', aspect='auto')
axes[2].set_title('Heatmap View\n(28×28 = 784 pixels)', fontsize=12, fontweight='bold')
axes[2].set_xlabel('Width (28 pixels)')
axes[2].set_ylabel('Height (28 pixels)')
plt.colorbar(im, ax=axes[2], label='Pixel Value')

plt.tight_layout()
plt.show()

# Show dimension distribution
print("\n=== Dimensionality Breakdown ===")
print(f"Original image: 28 rows × 28 columns = 784 pixels")
print(f"Each pixel = 1 dimension in the feature space")
print(f"Total feature vector: 784-dimensional space")
print(f"\nFor machine learning:")
print(f"  - Input: 784-dimensional vector (one per image)")
print(f"  - Each dimension: grayscale value (0-255)")
print(f"  - High-dimensional space requires dimensionality reduction")
print(f"  - Techniques: PCA, t-SNE, UMAP reduce 784D → 2D for visualization")
```
