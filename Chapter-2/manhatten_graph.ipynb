```python
import numpy as np
import matplotlib.pyplot as plt

# Define the two points
u = np.array([1, 2])
v = np.array([4, 6])

# Calculate Manhattan distance
manhattan_distance = np.sum(np.abs(u - v))

# Plot the points
plt.figure(figsize=(8, 6))
plt.scatter(*u, color='blue', s=100, label='u = [1, 2]')
plt.scatter(*v, color='red', s=100, label='v = [4, 6]')

# Draw Manhattan path: horizontal then vertical
plt.plot([u[0], v[0]], [u[1], u[1]], 'k--', linewidth=2)  # horizontal
plt.plot([v[0], v[0]], [u[1], v[1]], 'k--', linewidth=2)  # vertical

# Annotate the points
plt.text(u[0]-0.3, u[1]-0.3, 'u', fontsize=14, color='blue', weight='bold')
plt.text(v[0]+0.1, v[1]+0.1, 'v', fontsize=14, color='red', weight='bold')

# Annotate the distance formula
plt.text(1.5, 5, r"$d(u, v) = |1-4| + |2-6| = 3 + 4 = 7$", fontsize=14, color='black')

plt.title('Manhattan Distance Between Two Points')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True, alpha=0.3)
plt.legend()
plt.axis('equal')
```
