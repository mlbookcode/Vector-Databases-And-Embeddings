```python
import matplotlib.pyplot as plt
import numpy as np

def plot_vector_addition():
    """
    Visualize vector addition: v + w
    v = [2, 3], w = [1, -1]
    Result: v + w = [3, 2]
    """
    # Define vectors
    v = np.array([2, 3])
    w = np.array([1, -1])
    result = v + w
    
    # Create plot
    fig, ax = plt.subplots(1, 1, figsize=(10, 8))
    
    # Plot vector v (from origin)
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
              color='blue', width=0.008, label=f'v = [{v[0]}, {v[1]}]')
    
    # Plot vector w (from tip of v - geometric interpretation)
    ax.quiver(v[0], v[1], w[0], w[1], angles='xy', scale_units='xy', scale=1,
              color='red', width=0.008, label=f'w = [{w[0]}, {w[1]}]')
    
    # Plot resultant vector (from origin to final point)
    ax.quiver(0, 0, result[0], result[1], angles='xy', scale_units='xy', scale=1,
              color='green', width=0.01, label=f'v + w = [{result[0]}, {result[1]}]')
    
    # Mark key points
    ax.scatter(0, 0, color='black', s=100, zorder=5, label='Origin')
    ax.scatter(v[0], v[1], color='blue', s=100, zorder=5, label=f'Tip of v ({v[0]}, {v[1]})')
    ax.scatter(result[0], result[1], color='green', s=150, zorder=5, 
               label=f'Final point ({result[0]}, {result[1]})')
    
    # Add annotations
    ax.annotate(f'v = [{v[0]}, {v[1]}]', xy=(v[0]/2, v[1]/2), xytext=(v[0]/2-0.3, v[1]/2+0.3),
                fontsize=12, color='blue', fontweight='bold')
    ax.annotate(f'w = [{w[0]}, {w[1]}]', xy=(v[0]+w[0]/2, v[1]+w[1]/2), 
                xytext=(v[0]+w[0]/2+0.2, v[1]+w[1]/2-0.3),
                fontsize=12, color='red', fontweight='bold')
    ax.annotate(f'v + w = [{result[0]}, {result[1]}]', 
                xy=(result[0]/2, result[1]/2), xytext=(result[0]/2-0.5, result[1]/2-0.5),
                fontsize=12, color='green', fontweight='bold')
    
    # Set up plot
    ax.set_xlim(-1, 4)
    ax.set_ylim(-2, 4)
    ax.grid(True, alpha=0.3)
    ax.set_xlabel('X-axis', fontsize=12)
    ax.set_ylabel('Y-axis', fontsize=12)
    ax.set_title('Vector Addition: v + w = [2,3] + [1,-1] = [3,2]', 
                 fontsize=14, fontweight='bold')
    ax.legend(loc='upper left')
    ax.set_aspect('equal')
    
    plt.tight_layout()
    plt.show()


def plot_vector_sub():
    """
    Visualize vector addition: v + w
    v = [2, 3], w = [1, -1]
    Result: v + w = [3, 2]
    """
    # Define vectors
    v = np.array([2, 3])
    w = np.array([1, -1])
    result = v - w
    
    # Create plot
    fig, ax = plt.subplots(1, 1, figsize=(10, 8))
    
    # Plot vector v (from origin)
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
              color='blue', width=0.008, label=f'v = [{v[0]}, {v[1]}]')
    
    # Plot vector w (from tip of v - geometric interpretation)
    ax.quiver(v[0], v[1], w[0], w[1], angles='xy', scale_units='xy', scale=1,
              color='red', width=0.008, label=f'w = [{w[0]}, {w[1]}]')
    
    # Plot resultant vector (from origin to final point)
    ax.quiver(0, 0, result[0], result[1], angles='xy', scale_units='xy', scale=1,
              color='green', width=0.01, label=f'v - w = [{result[0]}, {result[1]}]')
    
    # Mark key points
    ax.scatter(0, 0, color='black', s=100, zorder=5, label='Origin')
    ax.scatter(v[0], v[1], color='blue', s=100, zorder=5, label=f'Tip of v ({v[0]}, {v[1]})')
    ax.scatter(result[0], result[1], color='green', s=150, zorder=5, 
               label=f'Final point ({result[0]}, {result[1]})')
    
    # Add annotations
    ax.annotate(f'v = [{v[0]}, {v[1]}]', xy=(v[0]/2, v[1]/2), xytext=(v[0]/2-0.3, v[1]/2+0.3),
                fontsize=12, color='blue', fontweight='bold')
    ax.annotate(f'w = [{w[0]}, {w[1]}]', xy=(v[0]-w[0]/2, v[1]-w[1]/2), 
                xytext=(v[0]-w[0]/2+0.2, v[1]-w[1]/2-0.3),
                fontsize=12, color='red', fontweight='bold')
    ax.annotate(f'v + w = [{result[0]}, {result[1]}]', 
                xy=(result[0]/2, result[1]/2), xytext=(result[0]/2-0.5, result[1]/2-0.5),
                fontsize=12, color='green', fontweight='bold')
    
    # Set up plot
    ax.set_xlim(-1, 4)
    ax.set_ylim(-2, 4)
    ax.grid(True, alpha=0.3)
    ax.set_xlabel('X-axis', fontsize=12)
    ax.set_ylabel('Y-axis', fontsize=12)
    ax.set_title('Vector Subtraction: v - w = [2,3] - [1,-1] = [3,2]', 
                 fontsize=14, fontweight='bold')
    ax.legend(loc='upper left')
    ax.set_aspect('equal')
    
    plt.tight_layout()
    plt.show()


    
    # Print calculation
    print("Vector Addition:")
    print(f"v = [{v[0]}, {v[1]}]")
    print(f"w = [{w[0]}, {w[1]}]")
    print(f"v + w = [{v[0]}+{w[0]}, {v[1]}+({w[1]})] = [{result[0]}, {result[1]}]")
    print("-" * 50)

def plot_scalar_multiplication():
    """
    Visualize scalar multiplication: 2v
    v = [2, 3]
    Result: 2v = [4, 6]
    """
    # Define vector and scalar
    v = np.array([2, 3])
    scalar = 2
    result = scalar * v
    
    # Create plot
    fig, ax = plt.subplots(1, 1, figsize=(10, 8))
    
    # Plot original vector v
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
              color='blue', width=0.008, label=f'v = [{v[0]}, {v[1]}]')
    
    # Plot scaled vector 2v
    ax.quiver(0, 0, result[0], result[1], angles='xy', scale_units='xy', scale=1,
              color='red', width=0.01, label=f'{scalar}v = [{result[0]}, {result[1]}]')
    
    # Mark key points
    ax.scatter(0, 0, color='black', s=100, zorder=5, label='Origin')
    ax.scatter(v[0], v[1], color='blue', s=100, zorder=5, 
               label=f'v endpoint ({v[0]}, {v[1]})')
    ax.scatter(result[0], result[1], color='red', s=150, zorder=5,
                label=f'2v endpoint ({result[0]}, {result[1]})')
    
    # Add magnitude comparison lines
    magnitude_v = np.sqrt(v[0]**2 + v[1]**2)
    magnitude_2v = np.sqrt(result[0]**2 + result[1]**2)
    
    # Add annotations
    ax.annotate(f'v = [{v[0]}, {v[1]}]\n||v|| = {magnitude_v:.2f}', 
                xy=(v[0], v[1]), xytext=(v[0]+0.2, v[1]+0.3),
                fontsize=12, color='blue', fontweight='bold',
                bbox=dict(boxstyle="round,pad=0.3", facecolor="lightblue", alpha=0.7))
    
    ax.annotate(f'2v = [{result[0]}, {result[1]}]\n||2v|| = {magnitude_2v:.2f}', 
                xy=(result[0], result[1]), xytext=(result[0]+0.2, result[1]+0.3),
                fontsize=12, color='red', fontweight='bold',
                bbox=dict(boxstyle="round,pad=0.3", facecolor="lightcoral", alpha=0.7))
    
    # Set up plot
    ax.set_xlim(-1, 7)
    ax.set_ylim(-1, 7)
    ax.grid(True, alpha=0.3)
    ax.set_xlabel('X-axis', fontsize=12)
    ax.set_ylabel('Y-axis', fontsize=12)
    ax.set_title('Scalar Multiplication: 2v = 2[2,3] = [4,6]', 
                 fontsize=14, fontweight='bold')
    ax.legend(loc='upper left')
    ax.set_aspect('equal')
    
    plt.tight_layout()
    plt.show()
    
    # Print calculation
    print("Scalar Multiplication:")
    print(f"v = [{v[0]}, {v[1]}]")
    print(f"scalar = {scalar}")
    print(f"{scalar}v = [{scalar}×{v[0]}, {scalar}×{v[1]}] = [{result[0]}, {result[1]}]")
    print(f"Original magnitude: ||v|| = {magnitude_v:.2f}")
    print(f"Scaled magnitude: ||2v|| = {magnitude_2v:.2f}")
    print(f"Scale factor: {magnitude_2v/magnitude_v:.1f}x")
    print("-" * 50)

def plot_vector_subtraction():
    """
    Visualize vector subtraction: v - w
    v = [5, 3], w = [2, 1]
    Result: v - w = [3, 2]
    """
    # Define vectors
    v = np.array([5, 3])
    w = np.array([2, 1])
    result = v - w
    
    # Create plot
    fig, ax = plt.subplots(1, 1, figsize=(10, 8))
    
    # Plot vector v (from origin)
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
              color='blue', width=0.008, label=f'v = [{v[0]}, {v[1]}]')
    
    # Plot vector w (from origin)
    ax.quiver(0, 0, w[0], w[1], angles='xy', scale_units='xy', scale=1,
              color='red', width=0.008, label=f'w = [{w[0]}, {w[1]}]')
   
    # Plot resultant vector v - w (from tip of w to tip of v)
    ax.quiver(w[0], w[1], result[0], result[1], angles='xy', scale_units='xy', scale=1,
              color='green', width=0.01, label=f'v - w = [{result[0]}, {result[1]}]')
    
    # Alternative: show v - w from origin
    ax.quiver(0, 0, result[0], result[1], angles='xy', scale_units='xy', scale=1,
              color='purple', width=0.008, alpha=0.8, 
              label=f'v - w (from origin)')
    
    # Mark key points
    ax.scatter(0, 0, color='black', s=100, zorder=5, label='Origin')
    ax.scatter(v[0], v[1], color='blue', s=100, zorder=5, 
               label=f'v endpoint ({v[0]}, {v[1]})')
    ax.scatter(w[0], w[1], color='red', s=100, zorder=5,
               label=f'w endpoint ({w[0]}, {w[1]})')
    ax.scatter(result[0], result[1], color='green', s=150, zorder=5,
               label=f'v-w endpoint ({result[0]}, {result[1]})')
    
    # Add annotations
    ax.annotate(f'v = [{v[0]}, {v[1]}]', xy=(v[0], v[1]), 
                xytext=(v[0]+0.2, v[1]+0.2),
                fontsize=12, color='blue', fontweight='bold')
    ax.annotate(f'w = [{w[0]}, {w[1]}]', xy=(w[0], w[1]), 
                xytext=(w[0]-0.5, w[1]+0.3),
                fontsize=12, color='red', fontweight='bold')
    ax.annotate(f'v - w = [{result[0]}, {result[1]}]', 
                xy=(w[0]+result[0]/2, w[1]+result[1]/2), 
                xytext=(w[0]+result[0]/2+0.3, w[1]+result[1]/2+0.3),
                fontsize=12, color='green', fontweight='bold')
    
    # Set up plot
    ax.set_xlim(-1, 6)
    ax.set_ylim(-1, 4)
    ax.grid(True, alpha=0.3)
    ax.set_xlabel('X-axis', fontsize=12)
    ax.set_ylabel('Y-axis', fontsize=12)
    ax.set_title('Vector Subtraction: v - w = [5,3] - [2,1] = [3,2]', 
                 fontsize=14, fontweight='bold')
    ax.legend(loc='upper left')
    ax.set_aspect('equal')
    
    plt.tight_layout()
    plt.show()
    
    # Print calculation
    print("Vector Subtraction:")
    print(f"v = [{v[0]}, {v[1]}]")
    print(f"w = [{w[0]}, {w[1]}]")
    print(f"v - w = [{v[0]}-{w[0]}, {v[1]}-{w[1]}] = [{result[0]}, {result[1]}]")
    print("-" * 50)

def plot_all_operations_combined():
    """
    Show all three operations in one comprehensive plot
    """
    fig, axes = plt.subplots(1, 3, figsize=(18, 6))
    
    # Vector Addition subplot
    v1 = np.array([2, 3])
    w1 = np.array([1, -1])
    result1 = v1 + w1
    
    axes[0].quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1,
                   color='blue', width=0.008)
    axes[0].quiver(v1[0], v1[1], w1[0], w1[1], angles='xy', scale_units='xy', scale=1,
                   color='red', width=0.008)
    axes[0].quiver(0, 0, result1[0], result1[1], angles='xy', scale_units='xy', scale=1,
                   color='green', width=0.01)
    axes[0].scatter([0, v1[0], result1[0]], [0, v1[1], result1[1]], 
                    colors=['black', 'blue', 'green'], s=80)
    axes[0].set_xlim(-1, 4)
    axes[0].set_ylim(-2, 4)
    axes[0].grid(True, alpha=0.3)
    axes[0].set_title('Addition: v + w = [3,2]', fontweight='bold')
    axes[0].set_aspect('equal')
    
    # Scalar Multiplication subplot
    v2 = np.array([2, 3])
    result2 = 2 * v2
    
    axes[1].quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1,
                   color='blue', width=0.008)
    axes[1].quiver(0, 0, result2[0], result2[1], angles='xy', scale_units='xy', scale=1,
                   color='red', width=0.01)
    axes[1].scatter([0, v2[0], result2[0]], [0, v2[1], result2[1]], 
                    colors=['black', 'blue', 'red'], s=80)
    axes[1].set_xlim(-1, 7)
    axes[1].set_ylim(-1, 7)
    axes[1].grid(True, alpha=0.3)
    axes[1].set_title('Scalar Mult: 2v = [4,6]', fontweight='bold')
    axes[1].set_aspect('equal')
    
    # Vector Subtraction subplot
    v3 = np.array([5, 3])
    w3 = np.array([2, 1])
    result3 = v3 - w3
    
    axes[2].quiver(0, 0, v3[0], v3[1], angles='xy', scale_units='xy', scale=1,
                   color='blue', width=0.008)
    axes[2].quiver(0, 0, w3[0], w3[1], angles='xy', scale_units='xy', scale=1,
                   color='red', width=0.008)
    axes[2].quiver(w3[0], w3[1], result3[0], result3[1], angles='xy', scale_units='xy', scale=1,
                   color='green', width=0.01)
    axes[2].scatter([0, v3[0], w3[0], result3[0]], [0, v3[1], w3[1], result3[1]], 
                    colors=['black', 'blue', 'red', 'green'], s=80)
    axes[2].set_xlim(-1, 6)
    axes[2].set_ylim(-1, 4)
    axes[2].grid(True, alpha=0.3)
    axes[2].set_title('Subtraction: v - w = [3,2]', fontweight='bold')
    axes[2].set_aspect('equal')
    
    plt.tight_layout()
    plt.show()

def main():
    """
    Main function to demonstrate all vector operations
    """
    print("2D Vector Operations Visualization")
    print("=" * 60)
    
    print("\n1. VECTOR ADDITION")
    print("=" * 30)
    plot_vector_addition()
    
    print("\n2. SCALAR MULTIPLICATION")
    print("=" * 30)
    plot_scalar_multiplication()
    
    print("\n3. VECTOR SUBTRACTION")
    print("=" * 30)
    plot_vector_subtraction()
    
    print("\n4. ALL OPERATIONS COMBINED")
    print("=" * 30)
    plot_all_operations_combined()
    
    print("\nAll vector operations visualized successfully!")

if __name__ == "__main__":
    main()
```
