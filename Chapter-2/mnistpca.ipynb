```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.datasets import fetch_openml
from sklearn.preprocessing import StandardScaler

# Load the MNIST dataset (70,000 samples of 28x28 images, flattened to 784 features)
mnist = fetch_openml('mnist_784', version=1, as_frame=False)
X = mnist.data                      # Feature matrix: each row is an image
y = mnist.target.astype(int)        # Labels: digits 0-9, converted to integers

# Standardize the data for PCA (zero mean, unit variance)
X_scaled = StandardScaler().fit_transform(X)

# Reduce the data to 2 principal components using PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled) # X_pca is now (n_samples, 2)

# Plot the 2D PCA projection (using only the first 5000 samples for speed)
plt.figure(figsize=(10, 8))
scatter = plt.scatter(
    X_pca[:5000, 0],                # First principal component
    X_pca[:5000, 1],                # Second principal component
    c=y[:5000],                     # Color by digit label
    cmap='tab10',                   # 10 distinct colors for digits 0-9
    alpha=0.6,                      # Slight transparency
    s=10                            # Small point size
)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA of MNIST Dataset (first 5000 samples)')
plt.colorbar(scatter, label='Digit Label') # Show color legend
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
