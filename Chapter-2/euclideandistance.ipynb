```python
import numpy as np
import matplotlib.pyplot as plt

# Define the two points
u = np.array([1, 2])
v = np.array([4, 6])

# Calculate Euclidean distance
distance = np.sqrt(np.sum((u - v) ** 2))

# Plot the points
plt.figure(figsize=(8, 6))
plt.scatter(*u, color='blue', s=100, label='u = [1, 2]')
plt.scatter(*v, color='red', s=100, label='v = [4, 6]')

# Draw the line between the points
plt.plot([u[0], v[0]], [u[1], v[1]], 'k--', linewidth=2, label=f'Distance = {distance:.1f}')

# Annotate the points
plt.text(u[0]-0.3, u[1]-0.3, 'u', fontsize=14, color='blue', weight='bold')
plt.text(v[0]+0.1, v[1]+0.1, 'v', fontsize=14, color='red', weight='bold')

# Annotate the distance formula
plt.text(1.5, 4.5, r"$d(u, v) = \sqrt{(4-1)^2 + (6-2)^2} = 5$", fontsize=14, color='black')

plt.title('Euclidean Distance Between Two Points')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True, alpha=0.3)
plt.legend()
plt.axis('equal')
plt.tight_layout()
```
